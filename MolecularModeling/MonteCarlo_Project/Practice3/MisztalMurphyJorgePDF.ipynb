## Required Task 6 - $E$ & $M$ Time Series at T=2.0, 2.27, 2.6

**Interpreting L=100 Graphs**

*  Following an initial equilibration period, the energy $E$ at each temperature is largely constant, though there are ~8-10% fluctuations. Higher fluctuations (about double) are seen at T=2.27, the critical temperature.
*  Magnetization:
   *  At T=2.0, the system is in the ferromagnetic phase and once equilibrated $M$ stays constant around $-9200$.
   *  At T=2.6, it's in the paramagnetic phase. Here, $M$ fluctuates around $0$ and with higher variance than the ferromagnetic phase.
   *  At T=2.27, the system is at the critical temperature where phase transition occurs. Here, $M$ fluctuates with the highest variance (between $\pm7500$) and even changes sign (phase transition).

**Comparing $M$ for L=20 vs L=100 at T=2.0**

*  Both systems are in the ferromagnetic phase at T=2.0, but the L=100 system has a lower magnetization than the L=20 system. This is because the L=100 system is 25x larger (10000 vs 400), though each system shows the same _proportional_ magnetization (~92%) to its respective system size.

## Required Task 7 - Plotting $\langle E \rangle$ & $\langle |M| \rangle$ with Statistical Error

**Interpretation** 

*  What we can observe in these three plots is how the estimated statistical error for energy and magnetization varies with the bin size used in the binning analysis. As we increase the bin size, the error tends to stabilize, indicating that we have reached a bin size large enough to capture the temporal correlations in the data.

*  This is especially evident near the critical temperature ($T=2.27$), where fluctuations are more pronounced and the autocorrelation time is longer. Below the critical temperature, the system exhibits more ordered behavior, which is reflected in smaller errors and a faster stabilization of the error with bin size. Above the critical temperature, the system is in a disordered phase, and due to the rapid decay of correlations, the error tends to stabilize even faster

## Required Task 8 - Plotting $\langle E \rangle/N$ and $\langle |M| \rangle/N$ with Statistical Error as a Function of $T$

**Interpretation**

*  At the low temperature (T=2.0), the error grows up to bin size 11 and plateaus, both for energy & magnetization. This state shows the lowest energy and highest magnetization.
*  At the critical temperature (T=2.27), the convergence is prolonged. Mean absolute magnetization is around 0.5 as phase transitions can occur at this temperature.
*  At the high temperature (T=2.6), the error rate is more constant throughout bin size, indicating fast decorrelation. The magnetization is low (averages towards zero) in this disordered state due to being above the critical temperature.

```raw
Jackknife Plateau Error (approx): 15.097334 (using fewest blocks/largest block size)
Naïve Propagation Error:          3.245069
Ratio (Naive / Jackknife):        0.21
```

**Heat Capacity**

The Naïve error estimate is nearly 15 times larger than the Jackknife error for the largest bin size and over 50 times larger for the smallest bins.
We get these much smaller Jackknife errors as a result of including the correlation between $\langle E^2 \rangle$ and $\langle E \rangle^2$. The fluctuations cancel out resulting in much smaller (true) value.
In the Naïve Propagation we assume that $\langle E^2 \rangle$ and $\langle E \rangle$ fluctuate independently so we add their variances.

**Susceptibility**

Here the Naïve error estimate is in the middle of the Jackknife errors range. It can be around 5 times smaller (for large bins) or 5 times larger (for small bins). 
It overestimates the error by ignoring the noise cancellation (covariance) between $\langle M^2 \rangle$ and $\langle |M| \rangle^2$, but simultaneously underestimates the error by ignoring the massive reduction in effective sample size caused by critical slowing down.

For small block sizes, the covariance effect dominates (Jackknife < Naïve). For large block sizes (the correct physical limit), the critical slowing down dominates, driving the true uncertainty well above the Naïve prediction (Jackknife > Naïve).
